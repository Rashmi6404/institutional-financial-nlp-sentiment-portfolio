```python
# Cell 2: Load FinBERT and define robust news fetching functions
import datetime
import xml.etree.ElementTree as ET
import matplotlib.pyplot as plt
import pandas as pd
import requests
import seaborn as sns
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer
import yfinance as yf

# 1. Device configuration (CUDA GPU if available)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Executing pipeline on: {device}")

# 2. Load FinBERT Pre-trained Model and Tokenizer
MODEL_NAME = "ProsusAI/finbert"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME).to(
    device
)


def fetch_ticker_news(ticker_symbol: str, max_articles: int = 8):
  """Extracts live financial headlines with automatic RSS failover."""
  ticker_symbol = ticker_symbol.strip().upper()
  headlines = []

  # Primary source: yfinance API
  try:
    ticker = yf.Ticker(ticker_symbol)
    if ticker.news:
      for item in ticker.news[:max_articles]:
        title = item.get("title") or item.get("content", {}).get("title", "")
        if title:
          headlines.append(title)
  except Exception:
    pass

  # Secondary source: Google News RSS fallback (bypasses IP throttling)
  if not headlines:
    try:
      url = f"https://news.google.com/rss/search?q={ticker_symbol}+stock&hl=en-US&gl=US&ceid=US:en"
      resp = requests.get(url, timeout=10)
      root = ET.fromstring(resp.content)
      for item in root.findall(".//item")[:max_articles]:
        title = item.find("title").text
        if " - " in title:
          title = title.rsplit(" - ", 1)[0]
        headlines.append(title)
    except Exception:
      pass

  return [{"ticker": ticker_symbol, "title": h} for h in headlines]


def analyze_sentiment_batch(records: list):
  """Runs batched tensor inference via FinBERT and calculates Net Compound Scores."""
  if not records:
    return pd.DataFrame()

  df = pd.DataFrame(records)
  titles = df["title"].tolist()

  # Tokenize all headlines in a single batch
  inputs = tokenizer(
      titles,
      padding=True,
      truncation=True,
      max_length=128,
      return_tensors="pt",
  ).to(device)

  # Model forward pass
  with torch.no_grad():
    outputs = model(**inputs)
    probs = torch.nn.functional.softmax(outputs.logits, dim=-1)

  # FinBERT outputs: [positive, negative, neutral]
  labels = ["positive", "negative", "neutral"]
  pred_labels = []
  net_scores = []

  for p in probs.cpu():
    pos, neg, neu = p[0].item(), p[1].item(), p[2].item()
    pred_labels.append(labels[p.argmax().item()])
    # Net Directional Score: P(Positive) - P(Negative) -> Range [-1.0, +1.0]
    net_scores.append(pos - neg)

  df["sentiment"] = pred_labels
  df["net_score"] = net_scores
  return df
```

```python
# Cell 3: Multi-ticker comparative aggregator and charting function
def run_sector_benchmark(tickers_input: str):
  """Aggregates scores across multiple peer equities and returns tabular + visual output."""
  tickers = [t.strip().upper() for t in tickers_input.split(",") if t.strip()]
  if not tickers:
    return "Error: Please enter at least one ticker symbol.", None, None

  all_news = []
  for t in tickers:
    all_news.extend(fetch_ticker_news(t, max_articles=8))

  if not all_news:
    return "Could not retrieve news for the given tickers.", None, None

  # Run FinBERT scoring
  df_scored = analyze_sentiment_batch(all_news)

  # Group by Ticker for the Sector Leaderboard
  leaderboard = (
      df_scored.groupby("ticker")
      .agg(
          Avg_Sentiment=("net_score", "mean"),
          Articles_Count=("title", "count"),
          Bullish_Pct=("sentiment", lambda x: (x == "positive").mean() * 100),
          Bearish_Pct=("sentiment", lambda x: (x == "negative").mean() * 100),
          Neutral_Pct=("sentiment", lambda x: (x == "neutral").mean() * 100),
      )
      .reset_index()
  )

  # Sort by most positive to most negative
  leaderboard = leaderboard.sort_values(
      by="Avg_Sentiment", ascending=False
  ).reset_index(drop=True)
  leaderboard["Avg_Sentiment"] = leaderboard["Avg_Sentiment"].round(3)
  leaderboard["Bullish_Pct"] = leaderboard["Bullish_Pct"].round(1)
  leaderboard["Bearish_Pct"] = leaderboard["Bearish_Pct"].round(1)
  leaderboard["Neutral_Pct"] = leaderboard["Neutral_Pct"].round(1)

  # Generate Leaderboard Chart
  fig, ax = plt.subplots(figsize=(10, max(4, len(leaderboard) * 0.7)))
  palette = [
      "#2ca02c" if s > 0.05 else "#d62728" if s < -0.05 else "#7f7f7f"
      for s in leaderboard["Avg_Sentiment"]
  ]

  sns.barplot(
      x="Avg_Sentiment",
      y="ticker",
      data=leaderboard,
      palette=palette,
      hue="ticker",
      legend=False,
      ax=ax,
  )
  ax.axvline(0, color="black", linestyle="--", linewidth=1)
  ax.set_xlim(-1.0, 1.0)
  ax.set_title(
      "Sector Cross-Sectional Sentiment Ranking", fontsize=13, fontweight="bold"
  )
  ax.set_xlabel("Net Sentiment Score (-1.0 Bearish to +1.0 Bullish)")
  ax.set_ylabel("Ticker")
  plt.tight_layout()

  # Markdown Report Summary
  summary = (
      f"### 📊 Sector Sentiment Summary\n"
      f"- **Top Bullish Ticker:** `{leaderboard.iloc[0]['ticker']}` (`{leaderboard.iloc[0]['Avg_Sentiment']}` score)\n"
      f"- **Top Bearish Ticker:** `{leaderboard.iloc[-1]['ticker']}` (`{leaderboard.iloc[-1]['Avg_Sentiment']}` score)\n"
      f"- **Total Articles Analyzed:** `{len(df_scored)}` across `{len(tickers)}` companies"
  )

  return summary, leaderboard, fig
```

```python
# Cell 4: Deploy the interactive browser UI with public share link
import gradio as gr

demo = gr.Interface(
    fn=run_sector_benchmark,
    inputs=gr.Textbox(
        value="V, MA, AXP, PYPL",
        label="Enter Comma-Separated Tickers",
        placeholder="e.g. V, MA, AXP or NVDA, AMD, INTC, TSM",
    ),
    outputs=[
        gr.Markdown(label="Executive Summary"),
        gr.Dataframe(label="Sector Leaderboard Metrics"),
        gr.Plot(label="Relative Tone Comparison Chart"),
    ],
    title="🏦 Institutional Financial NLP Sentiment Benchmark",
    description="Analyze and rank real-time market sentiment across competing equities using FinBERT.",
)

demo.launch(share=True, debug=True)
```

### 📝 Project Summary:

This project implements an **Institutional Financial NLP Sentiment Benchmark** designed to analyze and rank real-time market sentiment across competing equities. It leverages a pre-trained **FinBERT model** (`ProsusAI/finbert`) for sentiment classification, categorizing financial news headlines into positive, negative, or neutral sentiments.

The system features a robust news fetching mechanism that combines the `yfinance` API with a **Google News RSS fallback**, allowing the retrieval of up to **8 articles per ticker** to ensure comprehensive coverage and resilience against API throttling. News headlines are then tokenized with a `max_length` of **128** before being fed into the FinBERT model, which runs on either a CUDA GPU or CPU.

Sentiment analysis generates **Net Sentiment Scores** ranging from **-1.0 (bearish) to +1.0 (bullish)**, along with Bullish, Bearish, and Neutral percentages for each company. These metrics are aggregated to create a **Sector Leaderboard**, providing a cross-sectional ranking of equities based on their average sentiment. A dynamic bar chart visually represents these rankings.

Finally, the entire application is deployed as an interactive web interface using **Gradio**, enabling users to easily input multiple comma-separated ticker symbols (e.g., `V, MA, AXP, PYPL`) and receive an executive summary, a detailed leaderboard dataframe, and a visual sentiment comparison chart in real-time.

### Step-by-Step Project Explanation:

This project is designed to perform real-time financial sentiment analysis on news headlines for multiple stock tickers, providing a comparative benchmark.

#### **Cell 1: Install Required Libraries**

*   This cell (`D1FN5-7kpKte`) uses `!pip install -q` to quietly install all necessary Python libraries. These include:
    *   `yfinance`: For fetching financial data, including news, from Yahoo Finance.
    *   `transformers`: For using pre-trained NLP models like FinBERT.
    *   `torch`: The underlying deep learning framework (PyTorch) used by `transformers`.
    *   `pandas`: For data manipulation and analysis.
    *   `matplotlib` and `seaborn`: For creating visualizations.
    *   `gradio`: For building an interactive web user interface.
    *   `requests`: For making HTTP requests, particularly for fetching news from RSS feeds.


#### **Cell 2: Load FinBERT and Define Core Functions**

*   This cell (`cT4gIszxpKwi`) sets up the core components for sentiment analysis and news retrieval.

    1.  **Device Configuration**: It first checks if a CUDA-enabled GPU is available and configures PyTorch to use it for faster computation; otherwise, it defaults to the CPU. The output `Executing pipeline on: cuda` confirms GPU usage.

    2.  **Load FinBERT Model**: It loads the pre-trained `ProsusAI/finbert` model and its corresponding tokenizer from the Hugging Face `transformers` library. FinBERT is specifically trained on financial text, making it well-suited for this task.

    3.  **`fetch_ticker_news(ticker_symbol, max_articles=8)` function**: This function is responsible for gathering financial news headlines for a given stock ticker.
        *   **Primary Source**: It first attempts to fetch news from `yfinance`. If successful, it collects up to `max_articles` (defaulting to 8) and returns their titles.
        *   **Secondary Source (Fallback)**: If `yfinance` fails or returns no news, it falls back to parsing **Google News RSS feeds**. This is a robust mechanism to ensure news data is obtained even if one source has issues, helping bypass potential IP throttling.
        *   It returns a list of dictionaries, each containing the `ticker` and the `title` of a news article.

    4.  **`analyze_sentiment_batch(records)` function**: This function takes a list of news records (from `fetch_ticker_news`) and performs batched sentiment analysis.
        *   It extracts all news `titles`.
        *   **Tokenization**: It uses the loaded FinBERT tokenizer to convert these titles into a format the model can understand (`inputs`), ensuring consistent padding and truncation to `max_length=128`.
        *   **Model Inference**: The `model(**inputs)` performs a forward pass, generating raw `outputs` (logits).
        *   **Softmax and Scoring**: `torch.nn.functional.softmax` converts the logits into probabilities for 'positive', 'negative', and 'neutral' sentiments. The function then calculates a **Net Directional Score** (`pos - neg`), which ranges from -1.0 (most bearish) to +1.0 (most bullish). It also determines the `sentiment` label based on the highest probability.
        *   It returns a `pandas.DataFrame` augmented with 'sentiment' and 'net_score' columns for each news article.

#### **Cell 3: Multi-Ticker Comparative Aggregator and Charting Function**

*   This cell (`NxDVEEvypP-7`) defines the central function `run_sector_benchmark(tickers_input)` that orchestrates the entire analysis.

    1.  **Ticker Processing**: It takes a comma-separated string of tickers (e.g., "V, MA, AXP") and converts them into a list of uppercase ticker symbols.

    2.  **News Collection**: It iterates through each ticker, calling `fetch_ticker_news` to get up to 8 articles per company, combining them into a single list `all_news`.

    3.  **Sentiment Scoring**: It passes `all_news` to `analyze_sentiment_batch` to get a `DataFrame` (`df_scored`) with sentiment scores for all retrieved articles.

    4.  **Leaderboard Aggregation**: It groups `df_scored` by `ticker` to calculate aggregated metrics for each company:
        *   `Avg_Sentiment`: The mean of the net sentiment scores.
        *   `Articles_Count`: The total number of articles analyzed for that ticker.
        *   `Bullish_Pct`, `Bearish_Pct`, `Neutral_Pct`: The percentage of articles falling into each sentiment category.
        *   The results are rounded and sorted by `Avg_Sentiment` to create a `leaderboard` DataFrame.

    5.  **Chart Generation**: It uses `matplotlib.pyplot` and `seaborn` to generate a horizontal bar chart named "Sector Cross-Sectional Sentiment Ranking".
        *   Each bar represents a ticker, and its length corresponds to the `Avg_Sentiment` score.
        *   The color of the bar (`palette`) is dynamically set to green for bullish (>0.05), red for bearish (<-0.05), and gray for neutral sentiments.
        *   A vertical line at 0.0 (`ax.axvline(0)`) clearly separates bullish from bearish sentiment.

    6.  **Summary Report**: It generates a markdown `summary` string that highlights the top bullish and bearish tickers and the total number of articles analyzed, providing a quick overview.

    7.  **Return Values**: The function returns the `summary` (markdown string), the `leaderboard` (pandas DataFrame), and the `fig` (matplotlib figure object).


#### **Cell 4: Deploy Interactive Browser UI with Gradio**

*   This cell (`qNkPl56dpQE7`) uses the `gradio` library to create and deploy an interactive web interface for the sentiment analysis tool.

    1.  **`gr.Interface`**: This is the core Gradio component. It takes:
        *   `fn=run_sector_benchmark`: The Python function to execute when the user interacts with the UI.
        *   `inputs`: Defines the user input component. Here, `gr.Textbox` is used, allowing users to enter comma-separated ticker symbols. It provides a default value (`V, MA, AXP, PYPL`) and a placeholder.
        *   `outputs`: Defines how the results from `run_sector_benchmark` will be displayed.
            *   `gr.Markdown`: For the `summary` string.
            *   `gr.Dataframe`: For the `leaderboard` DataFrame.
            *   `gr.Plot`: For the `fig` (matplotlib chart).
        *   `title` and `description`: Provide context and information about the application.

    2.  **`demo.launch(share=True, debug=True)`**: This command starts the Gradio web server.
        *   `share=True`: Generates a public, temporary URL (like `https://8aea8026e31acacf22.gradio.live`) that allows anyone to access your application, making it easy to share and demo.
        *   `debug=True`: Provides more verbose output in the Colab console, which is useful for debugging.

In essence, this cell takes the analytical pipeline developed in previous cells and wraps it in a user-friendly web application, making the tool accessible without needing to write or run Python code directly.
